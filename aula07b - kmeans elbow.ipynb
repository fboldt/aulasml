```python
from sklearn.datasets import load_iris
data = load_iris()
X = data.data
y = data.target
```

```python
from sklearn.cluster import KMeans

kmeans_k3 = KMeans(n_clusters=3).fit(X)
print(f'kmeans_k3.inertia_ = {kmeans_k3.inertia_:.2f}')
kmeans_k8 = KMeans(n_clusters=8).fit(X)
print(f'kmeans_k8.inertia_ = {kmeans_k8.inertia_:.2f}')
```

```python
import matplotlib.pyplot as plt

def plot_elbow(X, max_k=10):
  inertias = []
  ks = range(2, max_k+1)
  for k in ks:
    kmeans = KMeans(n_clusters=k).fit(X)
    inertias.append(kmeans.inertia_)
  plt.figure(figsize=(12,5))
  plt.plot(ks, inertias, marker='o')
  plt.xlabel('Number of clusters (k)')
  plt.ylabel('Inertia')
  plt.title('Elbow Method for Optimal k')
  plt.xticks(ks)
  plt.grid()
  plt.show()

plot_elbow(X, max_k=10)
```

```python
from sklearn.datasets import load_wine
X, y = load_wine(return_X_y=True)
plot_elbow(X, max_k=10)
```
