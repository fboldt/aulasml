```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
from sklearn.manifold import TSNE
from sklearn.preprocessing import StandardScaler

# Load a sample dataset (handwritten digits)
digits = load_digits()
X = digits.data
y = digits.target

# Optional: Standardize features for better performance
X_scaled = StandardScaler().fit_transform(X)

# Create and configure TSNE
tsne = TSNE(
    n_components=2,        # Reduce to 2D for visualization
    perplexity=30,         # Controls balance between local/global aspects
    learning_rate=200,     # Typical range: 10–1000
    max_iter=1000,           # Number of iterations
    random_state=42        # For reproducibility
)

# Fit and transform the data
X_embedded = tsne.fit_transform(X_scaled)

# Plot the results
plt.figure(figsize=(8, 6))
scatter = plt.scatter(X_embedded[:, 0], X_embedded[:, 1], c=y, cmap='tab10', s=10)
plt.colorbar(scatter, label="Digit Label")
plt.title("t-SNE Visualization of Digits Dataset")
plt.xlabel("t-SNE Component 1")
plt.ylabel("t-SNE Component 2")
plt.show()
```
